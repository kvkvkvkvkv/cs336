Positional Encodings -> sinusoidal, RoPRE
Normalization -> LAyerNorm, RMSNorm, QK norm, pre norm vs post norm
Attention -> full, sparse, local attention, group query attention, multi head attention
Recurrence/ state space models/linear attention -> mamba, gated deltanet
MLP -> sense, mixture of experts 
Shape -> hidden dimension, depth, number of heads, number of experts
